```python
InsClsSampler = Sampler()
path = PROJECT_DIR+"/data/BayesianNetwork/BN_EUEMr.XDSL"
InsClsSampler.Load_BN(path)

# List of node names and their values
lst_NOEUD = InsClsSampler.lst_NOEUD
LIST_Dict = InsClsSampler.LIST_Dict
LIST_Dict

dd = {}
for key, item in LIST_Dict.items():
    dd[key] = {"Dependance": [],
              "Option": list(item.values()),}

for arc in InsClsSampler.bn.arcs():
    src = InsClsSampler.bn.variable(arc[0]).name()
    dest = InsClsSampler.bn.variable(arc[1]).name()
    dd[dest]["Dependance"].append(src)

ddf = pd.DataFrame(dd).T
ddf.to_csv("./SamplerBN.csv", sep=";", encoding="latin-1")
ddf
```
