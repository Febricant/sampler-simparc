```python
# liste des variables à comparer du réseau bayésien
variables_a_comparer = InsClsSampler.lst_NOEUD

# Calcul des fréquences des variables du BN dans l'échantillon généré
frequencies = {}
for variable in variables_a_comparer:
    frequencies[variable] = df1[variable].value_counts(normalize=True).to_dict()
    #print(f"Frequencies for {variable}:\n{frequencies[variable]}\n")

# Calcul des fréquences des variables du BN dans le fihier .XDSL
bn_data_path = PROJECT_DIR+"/data/processed/bayesian_network/BN_EUEMr.XDSL"
# Chargement des données du BN avec pyagrum

bn = gum.loadBN(bn_data_path)
ie=gum.LazyPropagation(bn)
ie.makeInference()

#calcul des distributions
bn_frequencies = {}
for variable in variables_a_comparer:
    #print (ie.posterior(variable))
    bn_frequencies[variable] = ie.posterior(variable).topandas().reset_index().drop(columns=['level_0'], axis=1).set_index('').to_dict()[0]
    #print(f"BN Frequencies for {variable}:\n{bn_frequencies[variable]}\n")
```

```python
# display ploly for all variables
import plotly.graph_objects as go
from plotly.subplots import make_subplots
cols=8
num_vars = int(len(variables_a_comparer)/cols)
fig = make_subplots(rows=num_vars, cols=cols, subplot_titles=list(variables_a_comparer.keys()))
for i, variable in enumerate(variables_a_comparer):
    row = i // cols + 1
    col = i % cols + 1
    sample_freq = frequencies[variable]
    bn_freq = bn_frequencies[variable]
    
    # sort categories to have consistent x-axis
    categories = sorted(list(set(sample_freq.keys()).union(set(bn_freq.keys()))))
    sample_values = [sample_freq.get(cat, 0) for cat in categories]
    bn_values = [bn_freq.get(cat, 0) for cat in categories]
    
    if i==0:
        fig.add_trace(go.Bar(x=categories, y=sample_values, name='Tirage', marker_color='blue',showlegend=True, legendgroup='Tirage'), row=row, col=col)
        fig.add_trace(go.Bar(x=categories, y=bn_values, name='Référence', marker_color='orange', showlegend=True, legendgroup='Référence'), row=row, col=col)
    else:
        fig.add_trace(go.Bar(x=categories, y=sample_values, name='Tirage', marker_color='blue', showlegend=False, legendgroup='Tirage'), row=row, col=col)
        fig.add_trace(go.Bar(x=categories, y=bn_values, name='Référence', marker_color='orange', showlegend=False, legendgroup='Référence'), row=row, col=col)
    #adjuste space between subplot to avoid xlend overlap
    fig.update_layout(bargap=0.2)
    #add fancy text style
    fig.update_xaxes(tickangle=45, tickfont=dict(size=10))
    #update subplot_title font size
    fig.layout.annotations[i].font.size = 9

    # change xlabel font size
    fig.update_xaxes(title_font=dict(size=8), row=row, col=col)
    

fig.update_layout(legend=dict(x=1.05, y=1))    

fig.update_layout(height=300*num_vars, title_text="Comparaison d'une distribution des variables entre l'échantillon et le réseau bayésien")
fig.show()
```
