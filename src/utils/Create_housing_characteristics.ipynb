```python
import pandas as pd
import os
import json
FILE_DIR = os.getcwd()
PROJECT_DIR = os.path.abspath(FILE_DIR+ "/../"+"/../")
FOLDER_DST = os.path.join(PROJECT_DIR+ "data","processed","housing_characteristics")
dct_housing_characteristics = {}
pathcsv = os.path.join(PROJECT_DIR, "data","input","housing_characteristics","options_lookup.tsv")
# CHARGER LE FICHIER INITIAL options_lookup.tsv
lookup_init = pd.read_csv(pathcsv,
						  sep='\t',header=None,names=['Parameter Name','Option Name','Measure Dir','Measure Arg 1',
								 				      'Measure Arg 2','Measure Arg 3','Measure Arg 4','Measure Arg 5',
													  'Measure Arg 6','Measure Arg 7','Measure Arg 8','Measure Arg 9',
													  'Measure Arg 10','Measure Arg 11','Measure Arg 12','Measure Arg 13',
													  'Measure Arg 14','Measure Arg 15','Measure Arg 16','Measure Arg 17',
													  'Measure Arg 18','Measure Arg 19','Measure Arg 20','Measure Arg 21',
													  'Measure Arg 22','Measure Arg 23','Measure Arg 24','Measure Arg 25',
													  'Measure Arg 26','Measure Arg 27','Measure Arg 28','Measure Arg 29',
													  'Measure Arg 30','Measure Arg 31','Measure Arg 32','Measure Arg 33',
													  'Measure Arg 34','Measure Arg 35','Measure Arg 36','Measure Arg 37',
													  'Measure Arg 38','Measure Arg 39','Measure Arg 40','Measure Arg 41',
													  'Measure Arg 42','Measure Arg 43','Measure Arg 44','Measure Arg 45',
													  'Measure Arg 46','Measure Arg 47','Measure Arg 48'],
						  skiprows=1,index_col=False)

dct_lookup_init = lookup_init[lookup_init["Parameter Name"]=="HVAC Heating Efficiency"].to_dict(orient='records')
dct_final = {}
for dct_items in dct_lookup_init:
    dct_final[dct_items.get('Option Name')] = {}
    for k, v in dct_items.items():
        if k == "Parameter Name":
            pass
        elif k=="Option Name":
            pass
        elif k.startswith("Measure Arg"):
            if pd.isna(v):
                pass
            else:
                k2 = v.split("=")[0]
                v2 = v.split("=")[1]
                try:
                    v2 = float(v2)
                except :
                    pass
                dct_final[dct_items.get('Option Name')][k2] = v2 
for k,v in dct_final.items():
    if pd.isnull(k):
        print("dct_HVAC_Heating" +"[\"" + "None" + "\"]" + " = " + json.dumps(v))
    else:
	    print("dct_HVAC_Heating" + "[\"" + k + "\"]" + " = " + json.dumps(v))
```
